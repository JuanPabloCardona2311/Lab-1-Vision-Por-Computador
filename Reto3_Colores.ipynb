Definir las 6 regiones de interes

```python
# Diccionario con las regiones de interés
# Formato: nombre : (x1, y1, x2, y2)

regiones = {
    "ROI 1 - Cielo azul": (1400, 320, 1600, 520),

    "ROI 2 - Pared turquesa": (120, 1760, 360, 1920),

    "ROI 3 - Pared verde lima": (1940, 1900, 2100, 2020),

    "ROI 4 - Pared morada": (2480, 2000, 2720, 2200),

    "ROI 5 - Pared azul": (2440, 2600, 2680, 2720),

    # Región con mezcla de varios colores
    "ROI 6 - Mezcla de colores": (3200, 2800, 3800, 3200)
}
```

```python
plt.figure(figsize=(15, 10))

for i, (nombre, coordenadas) in enumerate(regiones.items()):

    x1, y1, x2, y2 = coordenadas

    # Extraer la región
    roi = img[y1:y2, x1:x2]

    # Convertir de BGR a RGB para mostrar correctamente con matplotlib
    roi_rgb = cv2.cvtColor(roi, cv2.COLOR_BGR2RGB)

    plt.subplot(2, 3, i + 1)
    plt.imshow(roi_rgb)
    plt.title(nombre)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

Dibujar las 6 regiones

```python
# Crear una copia para no modificar la imagen original
img_regiones = img.copy()

# Colores de los rectángulos en formato BGR
colores_rectangulos = [
    (255, 255, 255),  # Blanco
    (0, 255, 255),    # Amarillo
    (0, 255, 0),      # Verde
    (255, 255, 0),    # Cian
    (0, 0, 255),      # Rojo
    (255, 0, 255)     # Magenta
]

for i, (nombre, coordenadas) in enumerate(regiones.items()):

    x1, y1, x2, y2 = coordenadas

    # Dibujar rectángulo
    cv2.rectangle(
        img_regiones,
        (x1, y1),
        (x2, y2),
        colores_rectangulos[i],
        12
    )

    # Escribir el nombre de la región
    cv2.putText(
        img_regiones,
        f"ROI {i+1}",
        (x1, y1 - 20),
        cv2.FONT_HERSHEY_SIMPLEX,
        2,
        colores_rectangulos[i],
        5
    )

cv2_imshow(img_regiones)
```

Calcular media y desviacion

```python
resultados = []

for nombre, coordenadas in regiones.items():

    x1, y1, x2, y2 = coordenadas

    # Extraer región
    roi = img[y1:y2, x1:x2]

    # Calcular media de los canales B, G y R
    media = np.mean(roi, axis=(0, 1))

    # Calcular desviación estándar de los canales B, G y R
    desviacion = np.std(roi, axis=(0, 1))

    # Guardar resultados
    resultados.append([
        nombre,
        media[0], media[1], media[2],
        desviacion[0], desviacion[1], desviacion[2]
    ])

    # Mostrar resultados
    print("\n" + "=" * 50)
    print(nombre)

    print("\nMedia:")
    print("Canal B:", round(media[0], 2))
    print("Canal G:", round(media[1], 2))
    print("Canal R:", round(media[2], 2))

    print("\nDesviación estándar:")
    print("Canal B:", round(desviacion[0], 2))
    print("Canal G:", round(desviacion[1], 2))
    print("Canal R:", round(desviacion[2], 2))
```

Tabla de Resultados

```python
df_resultados = pd.DataFrame(
    resultados,
    columns=[
        "Región",
        "Media B",
        "Media G",
        "Media R",
        "Desviación B",
        "Desviación G",
        "Desviación R"
    ]
)

# Redondear a 2 decimales
df_resultados = df_resultados.round(2)

display(df_resultados)
```

## Análisis de Resultados y Conclusiones

**Sobre la lectura de la imagen y su forma**

La imagen seleccionada contiene una gran variedad de colores, lo cual la hace adecuada para realizar el análisis de los canales de color. Al ser leída mediante OpenCV con la función `cv2.imread()`, la imagen se representa como una matriz tridimensional, cuya forma (`shape`) indica su altura, ancho y número de canales.

La forma obtenida permite conocer las dimensiones de la imagen. Los tres canales corresponden a los componentes de color que utiliza OpenCV, almacenados en el orden **BGR (Blue, Green, Red)**. Por lo tanto, cada píxel de la imagen está compuesto por tres valores que representan la intensidad de azul, verde y rojo, normalmente en un rango de 0 a 255.

**Sobre la selección de las regiones de interés**

Se seleccionaron seis regiones de interés o **ROI (Region of Interest)** mediante inspección visual de la imagen. Las primeras cinco regiones fueron seleccionadas buscando zonas relativamente homogéneas y con colores visualmente diferentes: una región azul correspondiente al cielo, una región turquesa, una región verde, una región morada y una región azulada. La sexta región fue seleccionada intencionalmente con una mezcla de varios colores y elementos, con el objetivo de comparar sus resultados estadísticos con los de las regiones homogéneas.

La selección de regiones homogéneas fue importante porque permite analizar zonas donde los píxeles presentan valores de color relativamente similares. De esta forma, la media de los canales B, G y R puede representar mejor el color predominante de cada región. En cambio, en una región con mezcla de colores, los valores medios representan únicamente un promedio general de todos los colores presentes.

**Análisis de los valores medios de los canales**

La media de cada canal permite conocer la intensidad promedio de los componentes azul, verde y rojo dentro de cada región. Los resultados obtenidos muestran una relación entre los valores numéricos y la apariencia visual de los colores seleccionados.

En la **ROI 1**, correspondiente a una región azul del cielo, el canal azul presenta una participación importante en los valores medios. Esto es coherente con la apariencia visual de la región, ya que una mayor presencia relativa del componente azul contribuye a producir una tonalidad azulada.

En la **ROI 2**, correspondiente a una región turquesa, los canales azul y verde presentan una participación importante en comparación con el canal rojo. Esto concuerda con la composición del color turquesa, que puede formarse mediante una combinación principalmente de componentes azules y verdes.

En la **ROI 3**, correspondiente a una región verde, el canal verde presenta una mayor participación relativa. Esto permite relacionar directamente el valor medio del canal G con la apariencia verde observada visualmente.

En la **ROI 4**, correspondiente a una región morada, se observa una combinación importante de los componentes azul y rojo. Esta combinación es coherente con la tonalidad morada, mientras que el componente verde presenta una menor participación relativa.

En la **ROI 5**, correspondiente a una región azulada, el canal azul tiene una participación importante en la composición del color. Sin embargo, también se observa que una imagen real no necesita tener un único canal con valor máximo y los demás en cero para representar un color determinado. El color final depende de la combinación y de la relación entre las intensidades de los tres canales BGR.

Finalmente, en la **ROI 6**, que contiene una mezcla de varios colores, los valores medios representan un promedio general de todos los píxeles presentes en la región. Por esta razón, los valores obtenidos no representan necesariamente un único color visual claramente identificable, sino la combinación estadística de los diferentes colores presentes.

**Análisis de la desviación estándar**

La desviación estándar permite medir qué tanto varían los valores de los píxeles dentro de cada región. Una desviación estándar baja indica que los píxeles tienen valores similares y, por lo tanto, la región es más homogénea. Por el contrario, una desviación estándar alta indica una mayor variación entre los valores de los píxeles.

Las primeras cinco regiones fueron seleccionadas buscando superficies de color relativamente uniforme. Por esta razón, sus desviaciones estándar son generalmente menores que las de una región con múltiples colores, aunque pueden existir variaciones causadas por cambios de iluminación, sombras, textura o pequeñas diferencias en la tonalidad.

La **ROI 6**, seleccionada como una región no homogénea, presenta una mayor variación en los valores de los canales. Esto se debe a que contiene diferentes colores y elementos visuales. Como consecuencia, los píxeles presentan valores más dispersos respecto a su media y la desviación estándar aumenta.

Este resultado permite comprobar que la desviación estándar es una medida útil para evaluar la homogeneidad de una región. Mientras más similares sean los píxeles, menor será la desviación estándar. Mientras mayor sea la mezcla y variación de colores, mayor será la desviación estándar.

**Comparación entre regiones homogéneas y no homogéneas**

Al comparar las seis regiones, se puede observar que la media y la desviación estándar proporcionan información diferente y complementaria. La media permite conocer cuál es la intensidad promedio de cada componente B, G y R, lo que ayuda a relacionar los datos numéricos con el color visual de la región.

Por ejemplo, una mayor participación relativa del canal B puede producir tonalidades azuladas, una mayor participación del canal G puede producir tonalidades verdes y una combinación de valores importantes en los canales B y R puede generar tonalidades moradas. De igual manera, una combinación importante de B y G puede relacionarse con tonalidades turquesas.

Por otro lado, la desviación estándar no indica directamente cuál es el color de la región, sino qué tanto varían los valores de ese color dentro de la zona analizada. Esta diferencia se observa claramente al comparar las regiones homogéneas con la ROI 6, donde la mezcla de colores produce una mayor dispersión de los valores.
